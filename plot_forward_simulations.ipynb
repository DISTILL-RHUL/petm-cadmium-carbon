```python
# Imports
import json
from matplotlib import pyplot
```

```python
# Load data files

with open('./data/output/steady_state.json', 'r') as file:
    steady_state = json.load(file)

with open('./data/output/perturbation_down.json', 'r') as file:
    perturbation_down = json.load(file)

with open('./data/output/perturbation_up.json', 'r') as file:
    perturbation_up = json.load(file)
```

```python
# Plot
colours = {"steady":"#000000",
                            "perturb_down":"#2355a1",
                            "perturb_up":"#d1803e"}

figure,axes = pyplot.subplots(3,1,figsize=(5,5),sharex=True)
axes[0].plot(steady_state['time'],steady_state['organic_mass'],zorder=3,color=colours["steady"])
axes[0].plot(steady_state['time'],perturbation_down['organic_mass'],zorder=3,color=colours["perturb_down"])
axes[0].plot(steady_state['time'],perturbation_up['organic_mass'],zorder=3,color=colours["perturb_up"])

axes[1].plot(steady_state['time'],steady_state['seawater_δ114Cd'],zorder=3,color=colours["steady"])
axes[1].plot(steady_state['time'],perturbation_down['seawater_δ114Cd'],zorder=3,color=colours["perturb_down"])
axes[1].plot(steady_state['time'],perturbation_up['seawater_δ114Cd'],zorder=3,color=colours["perturb_up"])

axes[2].plot(steady_state['time'],steady_state['seawater_mass'],zorder=3,color=colours["steady"])
axes[2].plot(steady_state['time'],perturbation_down['seawater_mass'],zorder=3,color=colours["perturb_down"])
axes[2].plot(steady_state['time'],perturbation_up['seawater_mass'],zorder=3,color=colours["perturb_up"])

axes[0].invert_xaxis()

## Set y axis exponent to 0
axes[0].get_yaxis().get_major_formatter().set_useOffset(False)


axes[0].set_ylabel("Organic Flux\n(mol/yr)")
axes[1].set_ylabel("$\\delta^{114}$Cd$_{sw}$\n (‰)")
axes[2].set_ylabel("Seawater\n Cadmium\n Mass (mol)")

axes[2].set_xlabel("Time (Ma)")
# axes[0].set_ylim((-0.5,0.5))

figure.tight_layout()
pyplot.show()
```
